```python
class libro:
  def __init__(self, isbn, titulo, autor, categoria, precio):
    self.isbn = isbn
    self.titulo = titulo
    self.autor = autor
    self.categoria = categoria
    self.precio = precio
    self.disponible = True # Added to be consistent with mostrar_info

  def mostrar_info(self):
    estado = "Disponible" if self.disponible else "No disponible"
    print(f"[{self.isbn}) {self.titulo} - {self.autor} - {self.categoria} - ${self.precio} {estado}]")

##############################################################################################################

class Estudiante:
  def __init__(self,num_control,nombre,carrera,fecha_nacimiento,telefono,correo):
    self.num_control = num_control
    self.nombre = nombre
    self.carrera = carrera
    self.fecha_nacimiento = fecha_nacimiento
    self.telefono = telefono
    self.correo = correo

  def mostrar_info(self): # Fixed typo: mostar_info -> mostrar_info
    print(f"[{self.num_control}] {self.nombre} - {self.carrera} | Nac: {self.fecha_nacimiento} | \
    Tel:{self.telefono} | correo: {self.correo}")


#################################################################################

class prestamo:
  contador_prestamos = 1


  def __init__(self, num_control, isbn, fecha_inicio, fecha_termino):
    self.folio = prestamo.contador_prestamos
    prestamo.contador_prestamos += 1
    self.num_control = num_control
    self.isbn = isbn
    self.fecha_inicio = fecha_inicio
    self.fecha_termino = fecha_termino

  def mostrar_info(self):
    print(f"Folio:{self.folio} | Estudiante: {self.num_control} | Libro: {self.isbn} \
    Del {self.fecha_inicio} al {self.fecha_termino}")


######################################################################################

class venta:
  contador_ventas = 1

  def __init__(self, isbn, num_control, fecha, cantidad, importe):
    self.folio = venta.contador_ventas
    venta.contador_ventas += 1
    self.isbn = isbn
    self.num_control = num_control # Initialized num_control
    self.fecha = fecha
    self.cantidad = cantidad
    self.importe = importe # Initialized importe

  def mostrar_info(self):
    print(f"Folio: {self.folio} | ISBN: {self.isbn} | Comprador: {self.num_control} | \
    Fecha: {self.fecha} | Cantidad: {self.cantidad} | Importe: ${self.importe}")

###################################################################################################

def alta_libro():
  print("\n--- Alta de Libro ---")
  isbn = input ("ISBN: ")
  for item_libro in libros: # Changed 'libro' to 'item_libro' to avoid shadowing the class name
    if item_libro.isbn == isbn:
      print("Ya existe un libro con ese ISBN!")
      return
  titulo = input ("Titulo: ")
  autor = input ("Autor: ")
  categoria = input ("Categoria: ")
  try:
      precio = float(input("Precio: "))
  except ValueError:
      print("Precio invalido")
      return
  nuevo_libro = libro(isbn, titulo, autor, categoria, precio)
  libros.append(nuevo_libro)
  print("libro registrado con exito")

####################################################################################################################
def alta_estudiante():
    print("\n--- Alta de Estudiante ---")
    num_control = input("Número de control: ")
    for est in estudiantes:
        if est.num_control == num_control:
            print("¡Ya existe un estudiante con ese número de control!")
            return
    nombre = input("Nombre: ")
    carrera = input("Carrera: ")
    fecha_nacimiento = input("Fecha de nacimiento (dd/mm/aaaa): ")
    telefono = input("Teléfono: ")
    correo = input("Correo electrónico: ")
    nuevo_estudiante = Estudiante(num_control, nombre, carrera, fecha_nacimiento, telefono, correo)
    estudiantes.append(nuevo_estudiante)
    print("Estudiante registrado exitosamente.")
####################################################################################################################
def consultar_libro():
    print("\n--- Consultar Libro por ISBN ---")
    isbn = input("Ingrese el ISBN del libro: ")
    for libro in libros:
        if libro.isbn == isbn:
            libro.mostrar_info()
            return
    print("Libro no encontrado.")

def consultar_estudiante():
    print("\n--- Consultar Estudiante por Número de Control ---")
    num_control = input("Ingrese el número de control: ")
    for est in estudiantes:
        if est.num_control == num_control:
            est.mostrar_info()
            return
    print("Estudiante no encontrado.")
####################################################################################################################
def mostrar_todos_libros():
    print("\n--- Lista de Todos los Libros ---")
    if not libros:
        print("No hay libros registrados.")
    else:
        for libro in libros:
            libro.mostrar_info()

def mostrar_todos_estudiantes():
    print("\n--- Lista de Todos los Estudiantes ---")
    if not estudiantes:
        print("No hay estudiantes registrados.")
    else:
        for est in estudiantes:
            est.mostrar_info()
####################################################################################################################
def realizar_prestamo():
    print("\n--- Realizar Préstamo ---")
    num_control = input("Número de control del estudiante: ")
    isbn = input("ISBN del libro: ")

    # Busca al estudiante
    estudiante = next((e for e in estudiantes if e.num_control == num_control), None)
    if not estudiante:
        print("Estudiante no encontrado.")
        return

    # Busca el libro
    libro = next((l for l in libros if l.isbn == isbn), None)
    if not libro:
        print("Libro no encontrado.")
        return

    if libro.prestado:
        print("El libro ya está prestado.")
        return

    fecha_inicio = input("Fecha de inicio (dd/mm/aaaa):")
    fecha_termino = input("Fecha de término (dd/mm/aaaa):")

    prestamo = Prestamo(num_control, isbn, fecha_inicio, fecha_termino)
    prestamos.append(prestamo)
    libro.prestado = True
    print(f"Préstamo registrado con folio {prestamo.folio}.")
####################################################################################################################
def mostrar_prestamos():
  print("\n--- Mostrar Préstamos ---")
  if not prestamos:
    print("No hay prestamos registrados.")
  else:
    for prestamo in prestamos:
      prestamo.mostrar_info()

####################################################################################################################

def devolver_libro():
  print("\n--- Devolver Libro ---")
  try:
    folio = int(input("folio del prestamos a devolver: "))
  except ValueError:
    print("folio invalido")
    return
  prestamo = next ((p for p in prestamo if p.folio == folio), None)
  if not prestamo:
    print("prestamo no encontrado")
    return
  libro = next((l for l in libros if l.isbn == prestamo.isbn), None)
  if libro:
    libro.prestado = False
    prestamos.remove(prestamo)
    print("libro devuelto con exito")
  else:
    print("libro no encontrado")

####################################################################################################################

def regitrar_venta():
  print("\n--- Registrar Venta ---")
  isbn = input("ISBN del libro a vender: ")
  libro = next((l for l in libros if l.isbn == isbn), None)
  if not libro:
    print("libro no encontrado")
    return
  try:
    cantidad = int(input("Cantidad a vender (solo 1 por ahora): "))
    if cantidad != 1:
      print("Solo se puede vender 1 libro por vez")
      return
  except ValueError:
    print("cantidad invalida")
    return
  num_control = input("numero de control del comprador (999999 su es publico general): ")
  if num_control == "999999":
    estudiante = next((e for e in estudiantes if e.num_control == num_control), None)
    if not estudiante:
      print("estudiante no encontrado")
      return
  fecha= input("fecha de la venta (dd/mm/aaaa): ")
  importe = libro.precio * cantidad
  venta = Venta(isbn, num_control, fecha, cantidad, importe)
  ventas.append(venta)
  libros.remove(libro)
  print(f"venta registrada. Folio: {venta.folio_venta}")

####################################################################################################################

def mostrar_historial_ventas():
  print("\n--- Historial de Ventas ---")
  if not ventas:
    print("No hay ventas registradas.")
  else:
    for v in ventas:
      v.mostrar_info()

####################################################################################################################

def mostrar_menu():
  print("\n--- SISTEMAS DE LIBRERIA ESCOLAR ---")
  print("1. Dar de alta el libro")
  print("2. Dar de alta estudiante")
  print("3. Consultar libro por ISBN")
  print("4. Consultar estudiante por numero de control")
  print("5. Mostrar todos los libros")
  print("6. Mostrar todos los estudiantes")
  print("7. Realizar prestamo")
  print("8. Mostrar prestamo")
  print("9. Devolver Libro")
  print("10. Registar ventas")

def iniciar_sistema():
  while True:
    mostrar_menu()
    opcion = input ("seleccione una opcion: ")
    if opcion == "1":
        alta_libro()
    elif opcion == "2":
        alta_estudiante()
    elif opcion == "3":
        consultar_libro()
    elif opcion == "4":
        consultar_estudiante()
    elif opcion == "5":
        mostrar_todos_libros()
    elif opcion == "6":
        mostrar_todos_estudiantes()
    elif opcion == "7":
        realizar_prestamo()
    elif opcion == "8":
        mostar_prestamos()
    elif opcion == "9":
        devolver_libro()
    elif opcion == "10":
        registar_venta()
    elif opcion == "11":
        mostar_historial_ventas()
    elif opcion == "12":
        print("saliendo del sistema. ¡HASTA LUEGO!")
        break
    else:
      print("Opcion no valida. Intenta de nuevo")



############################################################################

# Moved these global variables outside the function scope
libros = []
estudiantes = []
prestamos = []
ventas = []

iniciar_sistema()
```
