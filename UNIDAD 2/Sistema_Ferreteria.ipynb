```python
class Producto:
    def __init__(self, codigo, nombre, categoria, precio_unitario):
        self.codigo = codigo
        self.nombre = nombre
        self.categoria = categoria
        self.precio_unitario = precio_unitario

    def mostrar_info(self):
        print(f"[{self.codigo}] {self.nombre} - Categoria: {self.categoria} - Precio: ${self.precio_unitario}")

##############################################################################################################

class Cliente:
    def __init__(self, id_cliente, nombre, telefono, correo):
        self.id_cliente = id_cliente
        self.nombre = nombre
        self.telefono = telefono
        self.correo = correo

    def mostrar_info(self):
        print(f"[{self.id_cliente}] {self.nombre} | Tel: {self.telefono} | Correo: {self.correo}")

##############################################################################################################

class Venta:
    contador_ventas = 1

    def __init__(self, id_cliente, codigo_producto, fecha, cantidad, total):
        self.folio = Venta.contador_ventas
        Venta.contador_ventas += 1
        self.id_cliente = id_cliente
        self.codigo_producto = codigo_producto
        self.fecha = fecha
        self.cantidad = cantidad
        self.total = total

    def mostrar_info(self):
        print(f"Folio: {self.folio} | Fecha: {self.fecha} | Cliente: {self.id_cliente} | Producto: {self.codigo_producto} | Cantidad: {self.cantidad} | Total: ${self.total}")

##############################################################################################################

class InventarioProducto:
    def __init__(self, producto_obj, cantidad):
        self.producto = producto_obj
        self.cantidad = cantidad
        self.vendidos = 0

    def disponibles(self):
        return self.cantidad - self.vendidos

    def vender(self, cant):
        if self.disponibles() >= cant:
            self.vendidos += cant
            return True
        return False

    def agregar_stock(self, cant):
        self.cantidad += cant

    def mostrar_estado(self):
        p = self.producto
        print(f"[{p.codigo}] {p.nombre} | Total en sistema: {self.cantidad} | Disponibles: {self.disponibles()} | Vendidos: {self.vendidos}")

##############################################################################################################

def alta_producto():
    print("\n--- Alta de Producto ---")
    codigo = input("Código del producto: ")
    for inv_item in inventarios:
        if inv_item.producto.codigo == codigo:
            print("¡Ya existe un producto con ese código!")
            return

    nombre = input("Nombre: ")
    categoria = input("Categoría: ")
    try:
        precio_unitario = float(input("Precio unitario: "))
    except ValueError:
        print("Precio inválido.")
        return
    try:
        cantidad = int(input("Cantidad inicial en inventario: "))
    except ValueError:
        print("Cantidad inválida.")
        return

    nuevo_producto = Producto(codigo, nombre, categoria, precio_unitario)
    inventario_item = InventarioProducto(nuevo_producto, cantidad)
    inventarios.append(inventario_item)
    print("Producto agregado con éxito al inventario.")

##############################################################################################################

def mostrar_inventario():
    print("\n--- Inventario de Productos ---")
    if not inventarios:
        print("El inventario está vacío.")
    else:
        for item in inventarios:
            item.mostrar_estado()

##############################################################################################################

def consultar_producto():
    print("\n--- Consultar Producto por Código ---")
    codigo = input("Ingrese el código del producto: ")
    for inv_item in inventarios:
        if inv_item.producto.codigo == codigo:
            inv_item.producto.mostrar_info()
            return
    print("Producto no encontrado en inventario.")

##############################################################################################################

def registrar_cliente():
    print("\n--- Registrar Cliente ---")
    id_cliente = input("ID de cliente (No usar 999999, es para público general): ")
    if id_cliente == "999999":
        print("ID reservado para público en general. Usa otro.")
        return

    for cli in clientes:
        if cli.id_cliente == id_cliente:
            print("¡Ya existe un cliente con ese ID!")
            return

    nombre = input("Nombre: ")
    telefono = input("Teléfono: ")
    correo = input("Correo electrónico: ")
    nuevo_cliente = Cliente(id_cliente, nombre, telefono, correo)
    clientes.append(nuevo_cliente)
    print("Cliente registrado exitosamente.")

##############################################################################################################

def consultar_cliente():
    print("\n--- Consultar Cliente por ID ---")
    id_cliente = input("Ingrese el ID del cliente: ")
    for cli in clientes:
        if cli.id_cliente == id_cliente:
            cli.mostrar_info()
            return
    print("Cliente no encontrado.")

##############################################################################################################

def mostrar_todos_clientes():
    print("\n--- Lista de Todos los Clientes ---")
    if not clientes:
        print("No hay clientes registrados.")
    else:
        for cli in clientes:
            cli.mostrar_info()

##############################################################################################################

def registrar_venta():
    print("\n--- Registrar Venta ---")
    codigo = input("Código del producto a vender: ")
    inventario_item = next((item for item in inventarios if item.producto.codigo == codigo), None)

    if not inventario_item:
        print("Producto no encontrado en inventario.")
        return

    try:
        cantidad_a_vender = int(input("Cantidad a vender: "))
        if cantidad_a_vender <= 0:
            print("La cantidad debe ser mayor a cero.")
            return
    except ValueError:
        print("Cantidad inválida.")
        return

    id_cliente = input("ID del comprador (999999 si es público en general): ")
    if id_cliente != "999999":
        cliente_existe = next((c for c in clientes if c.id_cliente == id_cliente), None)
        if not cliente_existe:
            print("Cliente no encontrado. Registre al cliente primero o use 999999.")
            return


    if not inventario_item.vender(cantidad_a_vender):
        print(f"Inventario insuficiente. Solo quedan {inventario_item.disponibles()} unidades disponibles.")
        return

    fecha = input("Fecha de la venta (dd/mm/aaaa): ")
    total = inventario_item.producto.precio_unitario * cantidad_a_vender

    nueva_venta = Venta(id_cliente, codigo, fecha, cantidad_a_vender, total)
    ventas.append(nueva_venta)

    print(f"Venta registrada con éxito. Folio: {nueva_venta.folio} | Total a pagar: ${total}")

##############################################################################################################

def mostrar_historial_ventas():
    print("\n--- Historial de Ventas ---")
    if not ventas:
        print("No hay ventas registradas.")
    else:
        for v in ventas:
            v.mostrar_info()

##############################################################################################################

def mostrar_menu():
    print("\n--- SISTEMA DE FERRETERÍA ---")
    print("1. Dar de alta un producto")
    print("2. Mostrar inventario de productos")
    print("3. Consultar producto por código")
    print("4. Registrar cliente")
    print("5. Consultar cliente por ID")
    print("6. Mostrar todos los clientes")
    print("7. Registrar una venta")
    print("8. Mostrar historial de ventas")
    print("9. Salir")

def iniciar_sistema():
    while True:
        mostrar_menu()
        opcion = input("Seleccione una opción: ")

        if opcion == "1":
            alta_producto()
        elif opcion == "2":
            mostrar_inventario()
        elif opcion == "3":
            consultar_producto()
        elif opcion == "4":
            registrar_cliente()
        elif opcion == "5":
            consultar_cliente()
        elif opcion == "6":
            mostrar_todos_clientes()
        elif opcion == "7":
            registrar_venta()
        elif opcion == "8":
            mostrar_historial_ventas()
        elif opcion == "9":
            print("Saliendo del sistema. ¡HASTA LUEGO!")
            break
        else:
            print("Opción no válida. Intente de nuevo.")

############################################################################
# Listas en memoria
inventarios = []
clientes = []
ventas = []

# Arranque del programa
iniciar_sistema()
```
