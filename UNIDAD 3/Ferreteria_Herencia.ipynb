```python


class Persona:
    def __init__(self, nombre, correo, direccion, telefono):
        self.nombre = nombre
        self.correo = correo
        self.direccion = direccion
        self.telefono = telefono

    def mostrar_info(self):
        print(f"Nombre: {self.nombre} | Correo: {self.correo} | Dir: {self.direccion} | Tel: {self.telefono}")
############################################################################################################

class Cliente(Persona):
    def __init__(self, nombre, correo, direccion, telefono, rfc):
        super().__init__(nombre, correo, direccion, telefono)
        self.rfc = rfc
    def mostrar_info(self):
        super().mostrar_info()
        print(f"RFC: {self.rfc}")

############################################################################################################
class Empleado(Persona):
    def __init__(self, nombre, correo, direccion, telefono, id_empleado, departamento, usuario, contrasena):
        super().__init__(nombre, correo, direccion, telefono)
        self.id_empleado = id_empleado
        self.departamento = departamento
        self.usuario = usuario
        self.contrasena = contrasena

    def mostrar_info(self):
        super().mostrar_info()
        print(f"ID Empleado: {self.id_empleado} | Departamento: {self.departamento}")

############################################################################################################
class Producto:
    def __init__(self, id_prod, nombre, precio, cantidad):
        self.id_prod = id_prod
        self.nombre = nombre
        self.precio = precio
        self.cantidad = cantidad

    def mostrar_producto(self):
        print(f"[{self.id_prod}] {self.nombre} - ${self.precio:.2f} (Stock: {self.cantidad})")

############################################################################################################
class SistemaFerreteria:
    def __init__(self):
        self.empleados = []
        self.clientes = []
        self.inventario = []
        self.empleado_actual = None
        self._cargar_datos_defecto()
############################################################################################################
    def _cargar_datos_defecto(self):
        admin = Empleado("Admin Principal", "admin@ferreteria.com", "Conocida", "555-0000",
                         "EMP-001", "Administración", "admin", "admin123")
        self.empleados.append(admin)

        self.clientes.append(Cliente("Juan Pérez", "juan@mail.com", "Calle 1", "555-1111", "JUAN123POO"))
        self.inventario.append(Producto("100", "Martillo", 150.0, 10))
        self.inventario.append(Producto("101", "Clavos (Caja)", 50.0, 20))

   ############################################################################################################
    def iniciar_sistema(self):
        while True:
            print(" SISTEMA DE FERRETERÍA - LOGIN")
            usuario_input = input("Usuario (Escriba salir para apagar) (password: admin): ")
            if usuario_input.lower() == 'salir':
                print("Apagando el sistema...")
                break
            pass_input = input("Contraseña (admin123): ")
            logeado = False
            for empleado in self.empleados:
                if empleado.usuario == usuario_input and empleado.contrasena == pass_input:
                    self.empleado_actual = empleado
                    logeado = True
                    print(f"\n Acceso concedido. Bienvenido(a), {empleado.nombre}!")
                    self.menu_principal()
                    break

            if not logeado:
                print(" Credenciales incorrectas. Intente nuevamente.")
############################################################################################################
    def menu_principal(self):
        while True:
            print(f"\n --- MENÚ PRINCIPAL ({self.empleado_actual.departamento}) ---")
            print("1. Realizar Venta")
            print("2. Registrar Nuevo Producto")
            print("3. Ver Inventario")
            print("4. Registrar Nuevo Cliente")
            print("5. Ver Directorio de Clientes")
            print("6. Cerrar Sesión")

            opcion = input("Seleccione una opción: ")

            if opcion == '1':
                self.realizar_venta()
            elif opcion == '2':
                self.registrar_producto()
            elif opcion == '3':
                self.ver_inventario()
            elif opcion == '4':
                self.registrar_cliente()
            elif opcion == '5':
                self.ver_clientes()
            elif opcion == '6':
                print("Cerrando sesión...\n")
                self.empleado_actual = None
                break
            else:
                print(" Opción no válida.")
############################################################################################################
    def realizar_venta(self):
        self.ver_inventario()
        if not self.inventario:
            return

        id_prod = input("\nIngrese el ID del producto a vender: ")

        producto_encontrado = None
        for p in self.inventario:
            if p.id_prod == id_prod:
                producto_encontrado = p
                break

        if not producto_encontrado:
            print(" Producto no encontrado.")
            return

        try:
            cantidad = int(input("Cantidad a vender: "))
            if cantidad <= 0 or cantidad > producto_encontrado.cantidad:
                print(" Cantidad inválida o insuficiente.")
                return

            print("\n¿A quién se realiza la venta?")
            print("1. Público en general")
            print("2. Cliente Registrado")
            tipo_cliente = input("Seleccione opción: ")

            total = producto_encontrado.precio * cantidad

            if tipo_cliente == '2':
                rfc = input("Ingrese el RFC del cliente (JUAN123POO): ")
                cliente_valido = False
                for c in self.clientes:
                    if c.rfc == rfc:
                        cliente_valido = True
                        print(f" Venta al cliente: {c.nombre}. ¡Se aplica 10% de descuento!")
                        total *= 0.90
                        break
                if not cliente_valido:
                    print(" Cliente no encontrado. Se cobrará como público general.")
            else:
                print(" Venta a público en general.")

            producto_encontrado.cantidad -= cantidad
            print(f" Total a cobrar: ${total:.2f}")
            print(" Venta realizada con éxito.")

        except ValueError:
            print(" Por favor ingrese un número válido.")

############################################################################################################

    def registrar_producto(self):
        print("\n--- REGISTRO DE PRODUCTO ---")
        id_p = input("ID del Producto: ")
        nom = input("Nombre del Producto: ")
        try:
            pre = float(input("Precio: "))
            cant = int(input("Cantidad en inventario: "))
            nuevo_producto = Producto(id_p, nom, pre, cant)
            self.inventario.append(nuevo_producto)
            print(" Producto registrado correctamente.")
        except ValueError:
            print(" Error: Precio o cantidad inválidos.")

    def ver_inventario(self):
        print("\n--- INVENTARIO ACTUAL ---")
        if not self.inventario:
            print("El inventario está vacío.")
        else:
            for p in self.inventario:
                p.mostrar_producto()

    def registrar_cliente(self):
        print("\n--- REGISTRO DE CLIENTE ---")
        nom = input("Nombre: ")
        correo = input("Correo: ")
        dir = input("Dirección: ")
        tel = input("Teléfono: ")
        rfc = input("RFC: ")

        nuevo_cliente = Cliente(nom, correo, dir, tel, rfc)
        self.clientes.append(nuevo_cliente)
        print(" Cliente registrado correctamente.")

    def ver_clientes(self):
        print("\n--- DIRECTORIO DE CLIENTES ---")
        if not self.clientes:
            print("No hay clientes registrados.")
        else:
            for c in self.clientes:
                c.mostrar_info()
                print("-" * 20)

if __name__ == "__main__":
    app = SistemaFerreteria()
    app.iniciar_sistema()
```
