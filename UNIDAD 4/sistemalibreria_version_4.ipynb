```python
from datetime import datetime
from abc import ABC, abstractmethod


###############################################
class Persona(ABC):
  def __init__(self, nombre, fecha_nacimiento, correo):
    self.nombre = nombre
    self.fecha_nacimiento = fecha_nacimiento
    self.correo = correo


  @abstractmethod
  def mostrar_info(self):
      """Método abstracto: las subclases deben redefinirlo."""
      pass
#########################################
###############################################
#########################################

class Registrable(ABC):
    @abstractmethod
    def registrar_actividad(self, accion):
        pass

#subclase trabajador
class Trabajador(Persona, Registrable):
    def __init__(self, num_trabajador, nombre, correo, fecha_nacimiento, departamento, usuario, contrasena):
        super().__init__(nombre, fecha_nacimiento, correo)
        self.num_trabajador = num_trabajador
        self.departamento = departamento
        self.usuario = usuario
        self.contrasena = contrasena
        self.actividades = []

    def mostrar_info(self):
        return f"[Trabajador] Usuario: {self.usuario}, Núm. Trabajador: {self.num_trabajador}, Nombre: {self.nombre}"


    def registrar_actividad(self, accion):
        fecha_actual = datetime.now().strftime("%d-%m-%Y %H:%M:%S")
        registro = f"El {fecha_actual} - {self.nombre} realizó: {accion}"
        self.actividades.append(registro)

###############################################
def alta_trabajador():
  print("\n--- Alta de Trabajador ---")
  num_trabajador = input("numero de trabajador: ")
  for t in trabajadores:
    if t.num_trabajador == num_trabajador:
      print("ya existe un trabajador con ese numero de trabajador")
      return
  nombre = input("Nombre: ")
  correo = input("Correo: ")
  fecha_nacimiento = input("Fecha de nacimiento (YYYY-MM-DD): ")
  departamento = input("Departamento: ")
  usuario = input("Usuario de acceso: ")
  contraseña = input("Contraseña: ")

  nuevo = Trabajador(num_trabajador, nombre, fecha_nacimiento, correo, departamento, usuario, contraseña)
  trabajadores.append(nuevo)
  print(f"trabajador {nombre} agregado correctamente. \n")
  trabajador_activo.registrar_actividad(f"Alta de trabajador ´{nombre}´")
###############################################

def mostrar_trabajadores():
  print("\n--- Lista de Trabajadores ---")
  if not trabajadores:
    print("No hay trabajadores registrados.")
  else:
    for t in trabajadores:
      print(t.mostrar_info())

###############################################

def iniciar_sesion():
  print("\n=== INICIO DE SESION ===")
  usuario = input("Usuario: ")
  contraseña = input("Contraseña: ")

  for t in trabajadores:
      if t.usuario == usuario and t.contrasena == contraseña: # Changed t.contraseña to t.contrasena
          print(f"Bienvenido, {t.nombre} ({t.departamento})")
          t.registrar_actividad("inicio de sesion")
          return t # Return the active worker
  print("usuario o contraseña incorrectos. \n")
  return None

def cerrar_sesion():
    """Cierra la sesion actual"""
    global trabajador_activo
    if trabajador_activo:
      trabajador_activo.registrar_actividad(f"cierre de sesion {trabajador_activo.nombre}")
      print(f"sesion cerrada,{trabajador_activo.nombre}.")
    trabajador_activo = None
###############################################

def mostrar_registro_actividad():
  print("\n--- Registro de Actividad en el sistema ---")
  if trabajador_activo and trabajador_activo.actividades:
    for act in trabajador_activo.actividades:
      print("*", act)
  else:
    print("No hay actividades registradas para el trabajador activo.")

###############################################

class Libro:
  def __init__(self, isbn, titulo, autor, categoria, precio):
    self.isbn = isbn
    self.titulo = titulo
    self.autor = autor
    self.categoria = categoria
    self.precio = precio


  def mostrar_info(self):
    print(f"[{self.isbn}) {self.titulo} - {self.autor} - {self.categoria} - ${self.precio}]")

##############################################################################################################

class Estudiante(Persona):
  def __init__(self,num_control,nombre,carrera,fecha_nacimiento,telefono,correo):
    super().__init__(nombre,fecha_nacimiento,correo) #hereda atributos de persona
    self.num_control = num_control
    self.carrera = carrera
    self.telefono = telefono

    def mostrar_info(self):
        # Sustituimos super().mostrar_info() por los atributos directos
        return (f"[Estudiante] Nombre: {self.nombre}, Correo: {self.correo}, "
                f"Nacimiento: {self.fecha_nacimiento}, Número control: {self.num_control}, "
                f"Carrera: {self.carrera}, Tel: {self.telefono}")

#################################################################################

class Prestamo:
  contador_prestamos = 1


  def __init__(self, num_control, isbn, fecha_inicio, fecha_termino):
    self.folio = Prestamo.contador_prestamos
    Prestamo.contador_prestamos += 1
    self.num_control = num_control
    self.isbn = isbn
    self.fecha_inicio = fecha_inicio
    self.fecha_termino = fecha_termino

  def mostrar_info(self):
    print(f"Folio:{self.folio} | Estudiante: {self.num_control} | Libro: {self.isbn}     Del {self.fecha_inicio} al {self.fecha_termino}")


######################################################################################

class Venta:
  contador_ventas = 1

  def __init__(self, isbn, num_control, fecha, cantidad, importe):
    self.folio = Venta.contador_ventas
    Venta.contador_ventas += 1
    self.isbn = isbn
    self.num_control = num_control
    self.fecha = fecha
    self.cantidad = cantidad
    self.importe = importe

  def mostrar_info(self):
    print(f"Folio: {self.folio} | ISBN: {self.isbn} | Comprador: {self.num_control} |     Fecha: {self.fecha} | Cantidad: {self.cantidad} | Importe: ${self.importe}")

###################################################################################################

def alta_libro():
  print("\n--- Alta de Libro ---")
  isbn = input ("ISBN: ")
  for inv_item in Inventarios:
    if inv_item.libro.isbn == isbn:
      print("Ya existe un libro con ese ISBN!")
      return
  titulo = input ("Titulo: ")
  autor = input ("Autor: ")
  categoria = input ("Categoria: ")
  try:
      precio = float(input("Precio: "))
  except ValueError:
      print("Precio invalido")
      return
  try:
      cantidad = int(input("Cantidad en inventario: "))
  except ValueError:
      print("cantidad invalida")
      return
  nuevo_libro = Libro(isbn, titulo, autor, categoria, precio)
  inventario_item = InventarioLibro(nuevo_libro, cantidad)
  Inventarios.append(inventario_item)
  print("libro agregado con exito al inventario")
  trabajador_activo.registrar_actividad(f"Alta de libro ´{titulo}´")
####################################################################################################################
def alta_estudiante():
    print("\n--- Alta de Estudiante ---")
    num_control = input("Número de control: ")
    for est in estudiantes:
        if est.num_control == num_control:
            print("¡Ya existe un estudiante con ese número de control!")
            return
    nombre = input("Nombre: ")
    carrera = input("Carrera: ")
    fecha_nacimiento = input("Fecha de nacimiento (dd/mm/aaaa): ")
    telefono = input("Teléfono: ")
    correo = input("Correo electrónico: ")
    nuevo_estudiante = Estudiante(num_control, nombre, carrera, fecha_nacimiento, telefono, correo)
    estudiantes.append(nuevo_estudiante)
    print("Estudiante registrado exitosamente.")
    trabajador_activo.registrar_actividad(f"Alta de estudiante ´{nombre}´")
####################################################################################################################
def consultar_libro():
    print("\n--- Consultar Libro por ISBN ---")
    isbn = input("Ingrese el ISBN del libro: ")
    for inv_item in Inventarios:
        if inv_item.libro.isbn == isbn:
            inv_item.libro.mostrar_info()
            return
    print("Libro no encontrado en inventario.")

def consultar_estudiante():
    print("\n--- Consultar Estudiante por Número de Control ---")
    num_control = input("Ingrese el número de control: ")
    for est in estudiantes:
        if est.num_control == num_control:
            est.mostrar_info()
            return
    print("Estudiante no encontrado.")
####################################################################################################################
def mostrar_inventario():
  print("\n--- Inventario de todos los libros ---")
  if not Inventarios:
    print("El inventario está vacío.")
  else:
    for item in Inventarios:
      item.mostrar_estado()

def mostrar_todos_estudiantes():
    print("\n--- Lista de Todos los Estudiantes ---")
    if not estudiantes:
        print("No hay estudiantes registrados.")
    else:
        for est in estudiantes:
            est.mostrar_info()
####################################################################################################################
def realizar_prestamo():
    print("\n--- Realizar Préstamo ---")
    num_control = input("Número de control del estudiante: ")
    estudiante = next((e for e in estudiantes if e.num_control == num_control), None)
    if not estudiante:
        print("Estudiante no encontrado.")
        return
    isbn = input("ISBN del libro: ")
    inventario_item = next((item for item in Inventarios if item.libro.isbn == isbn), None)
    if not inventario_item:
        print("Libro no encontrado en el inventario.")
        return

    if not inventario_item.prestar():
        print("El libro no está disponible para préstamo.")
        return

    fecha_inicio = input("Fecha de inicio (dd/mm/aaaa): ")
    fecha_termino = input("Fecha de término (dd/mm/aaaa): ")
    new_prestamo = Prestamo(num_control, isbn, fecha_inicio, fecha_termino)
    prestamos.append(new_prestamo)
    print(f"Préstamo registrado. Folio: {new_prestamo.folio}")
    trabajador_activo.registrar_actividad(f"Prestamo realizado ´{num_control}´")
####################################################################################################################
def mostrar_prestamos():
  print("\n--- Mostrar Préstamos ---")
  if not prestamos:
    print("No hay prestamos registrados.")
  else:
    for p in prestamos:
      p.mostrar_info()

####################################################################################################################
def mostrar_inventario():
  print("\n--- Inventario de todos los libros ---")
  if not Inventarios:
    print("inventario vacio")
  else:
    for item in Inventarios:
      item.mostrar_estado()

####################################################################################################################

def devolver_libro():
  print("\n--- Devolver Libro ---")
  try:
    folio = int(input("folio del prestamos a devolver: "))
  except ValueError:
    print("folio invalido")
    return

  prestamo_a_devolver = next ((p for p in prestamos if p.folio == folio), None)
  if not prestamo_a_devolver:
    print("prestamo no encontrado")
    return

  inventario_item = next((item for item in Inventarios if item.libro.isbn == prestamo_a_devolver.isbn), None)
  if inventario_item:
    inventario_item.devolver()
  prestamos.remove(prestamo_a_devolver)
  print("libro devuelto")

####################################################################################################################

def registrar_venta():
  print("\n--- Registrar Venta ---")
  isbn = input("ISBN del libro a vender: ")
  inventario_item = next((item for item in Inventarios if item.libro.isbn == isbn), None)
  if not inventario_item:
    print("libro no encontrado en inventario")
    return
  try:
    cantidad = int(input("Cantidad a vender (solo 1 por ahora): "))
    if cantidad != 1:
      print("Solo se puede vender 1 libro por vez")
      return
  except ValueError:
    print("cantidad invalida")
    return

  num_control = input("numero de control del comprador (999999 si es publico general): ")
  if num_control != "999999":
      estudiante = next((e for  e in estudiantes if e.num_control == num_control), None)
      if not estudiante:
        print("estudiante no encontrado")
        return

  if not inventario_item.vender():
      print("Libro no disponible para venta.")
      return

  fecha= input("fecha de la venta (dd/mm/aaaa): ")
  importe = inventario_item.libro.precio * cantidad
  new_venta = Venta(isbn, num_control, fecha, cantidad, importe)
  ventas.append(new_venta)

  print(f"venta registrada. Folio: {new_venta.folio}")

####################################################################################################################

def mostrar_historial_ventas():
  print("\n--- Historial de Ventas ---")
  if not ventas:
    print("No hay ventas registradas.")
  else:
    for v in ventas:
      v.mostrar_info()

####################################################################################################################

class InventarioLibro:
  def __init__(self, libro_obj, cantidad):
    self.libro = libro_obj
    self.total = cantidad
    self.prestados = 0
    self.vendidos = 0

  def disponible(self):
    return self.total - self.prestados - self.vendidos

  def prestar(self):
      if self.disponible() > 0:
        self.prestados += 1
        return True
      return False
  def devolver(self):
      if self.prestados > 0:
        self.prestados -= 1
        return True
      return False

  def vender(self):
    if self.disponible() > 0:
      self.vendidos += 1
      return True
    return False

  def mostrar_estado(self):
    l = self.libro
    print(f"[{l.isbn}] {l.titulo} - {l.autor} | Total: {self.total} |     Disponibles: {self.disponible()} | Prestado: {self.prestados} | Vendidos: {self.vendidos}")

####################################################################################################################


####################################################################################################################

def mostrar_menu():
  print("\n--- SISTEMAS DE LIBRERIA ESCOLAR (con inventario) ---")
  print("1. Dar de alta el libro")
  print("2. Dar de alta estudiante")
  print("3. Consultar libro por ISBN")
  print("4. Consultar estudiante por numero de control")
  print("5. Mostrar inventario")
  print("6. Mostrar todos los estudiantes")
  print("7. Realizar prestamo")
  print("8. Mostrar prestamos")
  print("9. Devolver Libro")
  print("10. Registrar ventas")
  print("11. Mostrar historial de ventas")
  print("12. Dar de alta trabajador")
  print("13. Mostrar todos los trabajadores")
  print("14. Cerrar sesion")
  print("15. mostrar registro de actividad en el sistema")
  print("16. Salir del sistema")

def iniciar_sistema():
  global trabajador_activo
  trabajador_activo = None # Initialize with no active worker

  while True: # Keep the system running indefinitely until explicitly exited
    if trabajador_activo is None: # If no one is logged in, prompt for login
      trabajador_activo = iniciar_sesion()
      if trabajador_activo is None: # If login fails, continue to re-prompt for login
        continue

    # If we are here, a worker is logged in
    mostrar_menu()
    opcion = input ("seleccione una opcion: ")

    if opcion == "1":
        alta_libro()
    elif opcion == "2":
        alta_estudiante()
    elif opcion == "3":
        consultar_libro()
    elif opcion == "4":
        consultar_estudiante()
    elif opcion == "5":
         mostrar_inventario()
    elif opcion == "6":
        mostrar_todos_estudiantes()
    elif opcion == "7":
        realizar_prestamo()
    elif opcion == "8":
        mostrar_prestamos()
    elif opcion == "9":
        devolver_libro()
    elif opcion == "10":
        regitrar_venta()
    elif opcion == "11":
        mostrar_historial_ventas()
    elif opcion == "12":
        alta_trabajador()
    elif opcion == "13":
        mostrar_trabajadores()
    elif opcion == "14":
        cerrar_sesion() # cerrar_sesion now only sets trabajador_activo to None
        # The next iteration of the while True loop will detect trabajador_activo is None
        # and prompt for a new login.
    elif opcion == "15":
        mostrar_registro_actividad()
    elif opcion == "16":
        print("saliendo del sistema")
        break # Exit the infinite loop
    else:
      print("Opcion no valida.")



############################################################################

Inventarios = []
estudiantes = []
prestamos = []
ventas = []
trabajadores = []

trabajador_activo = None

admin = Trabajador("0001", "administrador del sistema", "admin@gmail.com", "1980-01-01",
                   "sistemas", "admin","admin")

trabajadores.append(admin)


#==============
#Ejecucion Principal
#==============
if __name__ == "__main__":
  iniciar_sistema()

```
